```python
import pandas as pd
from pathlib import Path
```

```python
EVENT_LOG_DIR = "event_log"
EVENT_LOG_FILE = "curia_log_en.csv"
EVENT_LOG_PATH = Path(EVENT_LOG_DIR) / EVENT_LOG_FILE

CASEID_COL = "CaseID"
ACTIVITY_COL = "Activity"
TIMESTAMP_COL = "Timestamp"
```

```python
df_log = pd.read_csv(EVENT_LOG_PATH, sep=";", low_memory=False)
```

```python
df_log.head()
```

```python
print("Number of cases:", df_log[CASEID_COL].nunique())
```

```python
print("Activities list")
activity_list = df_log[ACTIVITY_COL].unique().tolist()
# print(activity_list)
for i, activity in enumerate(activity_list, start=1):
    print(f"{i}: {activity}")
```

```python
print("First activity list")
df_first_events = (
    df_log[[CASEID_COL, ACTIVITY_COL, TIMESTAMP_COL]]
    .copy()
    .assign(**{TIMESTAMP_COL: pd.to_datetime(df_log[TIMESTAMP_COL], errors="coerce")})
    .sort_values([CASEID_COL, TIMESTAMP_COL])
    .drop_duplicates(subset=[CASEID_COL], keep="first")
)

activity_list_first = sorted(
    df_first_events[ACTIVITY_COL]
    .dropna()
    .astype(str)
    .unique()
    .tolist()
)

for i, activity in enumerate(activity_list_first, start=1):
    print(f"{i}: {activity}")

print(f"Total distinct activities as the first activity in the case: {len(activity_list_first)}")
```

```python
print("Last activity list")
df_last_events = (
    df_log[[CASEID_COL, ACTIVITY_COL, TIMESTAMP_COL]]
    .copy()
    .assign(**{TIMESTAMP_COL: pd.to_datetime(df_log[TIMESTAMP_COL], errors="coerce")})
    .sort_values([CASEID_COL, TIMESTAMP_COL])
    .drop_duplicates(subset=[CASEID_COL], keep="last")
)

activity_list_last = sorted(
    df_last_events[ACTIVITY_COL]
    .dropna()
    .astype(str)
    .unique()
    .tolist()
)

for i, activity in enumerate(activity_list_last, start=1):
    print(f"{i}: {activity}")

print(f"Total distinct activities as the last activity in the case: {len(activity_list_last)}")
```

```python
print("Area of Law list")
area_of_law_list = df_log["Area_of_Law"].unique().tolist()
for i, area in enumerate(area_of_law_list, start=1):
    print(f"{i}: {area}")   
```
